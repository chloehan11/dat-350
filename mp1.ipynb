Mini-Project 1: Getting Data Driven Answers from Surveys

```python
import pandas as pd
```

```python
musicDF = pd.read_csv("music_habits.csv")

musicDF.head()
```

*I asked a couple extra questions to get a better idea of music habits, but in my report focused only focused on my main questions. Below includes data found from all questions.

```python
print(musicDF.columns)
```

Age range (percentages)

```python
ageCounts = musicDF["What is your age range?"].value_counts()

totalResponses = len(musicDF)

agePercentages = (ageCounts/totalResponses)*100
print(agePercentages)
```

Year in school (percentage)

```python
yearCounts = musicDF["What year are you in school?"].value_counts()

yearPercentages = (yearCounts/totalResponses)*100
print(yearPercentages)
```

Major category (percentage)

```python
majorCounts = musicDF["What is your major category? "].value_counts()

majorPercentages = (majorCounts/totalResponses)*100
print(majorPercentages)
```

What types of music do people listen to most often? (percentage)

```python
musicGenres = musicDF["What type(s) of music do you listen to most often?"].dropna().str.split(", ")

allGenres = musicGenres.explode()

genreCounts = allGenres.value_counts()

genrePercentages = (genreCounts/totalResponses)*100
print(genrePercentages)
```

When do people listen to music most often? (percentage)

```python
timeCounts = musicDF["When do you listen to music most often?"].value_counts()

timePercentages = (timeCounts/totalResponses)*100
print(timePercentages)
```

Why do people listen to music at that time of day? (percentages)

```python
whyTimeCounts = musicDF["Why do you listen to music at that time of day?"].value_counts()

whyTimePercentages = (whyTimeCounts/totalResponses)*100
print(whyTimePercentages)
```

What activities do people usually listen to music to? (percentages)

```python
activitiesList = musicDF["During which activities do you usually listen to music?"].dropna().str.split(", ")

allActivities = activitiesList.explode()

activitiesCount = allActivities.value_counts()

activitiesPercentage = (activitiesCount/totalResponses)*100
print(activitiesPercentage)
```

Where do people usually listen to music? (percentages)

```python
locationsList = musicDF["Where do you usually listen to music?"].dropna().str.split(", ")

allLocations = locationsList.explode()

locationsCount = allLocations.value_counts()

locationsPercentage = (locationsCount/totalResponses)*100
print(locationsPercentage)
```

Why do people usually listen to music in those locations? (percentages)

```python
whyLocCounts = musicDF["Why do you listen to music in those locations?"].value_counts()

whyLocPercentages = (whyLocCounts/totalResponses)*100
print(whyLocPercentages)
```

Why do people primarily listen to music? (percentages)

```python
whyListenCounts = musicDF["Why do you primarily listen to music?"].value_counts()

whyListenPercentages = (whyListenCounts/totalResponses)*100
print(whyListenPercentages)
```
