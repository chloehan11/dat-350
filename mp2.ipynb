```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
candyDF = pd.read_csv("candy-data.csv")
candyDF.head()
```

Clean dataframe

```python
candyDF.isna().sum()
```

```python
candyDF["sugarpercent"] = pd.to_numeric(candyDF["sugarpercent"])
candyDF["pricepercent"] = pd.to_numeric(candyDF["pricepercent"])
candyDF["winpercent"] = pd.to_numeric(candyDF["winpercent"])
```

```python
candyDF["chocolate"].value_counts()
```

Figure 1: Distribution of Popularity

```python
plt.hist(candyDF["winpercent"], bins=10, color = "#bc8ebf")
plt.xlabel("Win Percent (Popularity)")
plt.ylabel("Number of Candies")
plt.title("Distribution of Candy Popularity")
plt.show()
```

Figure 2: Chocolate vs Non-Chocolate Popularity

```python
chocAvg = candyDF.groupby("chocolate")["winpercent"].mean()
barColors = ["#b2adb3", "#6e4008"]

plt.bar(["Non-Chocolate", "Chocolate"], chocAvg, color = barColors)
plt.xlabel("Candy Type")
plt.ylabel("Average Win Percent")
plt.title("Average Popularity: Chocolate vs Non-Chocolate")
plt.show()
```

Figure 3: Ingredient Comparison

```python
chocolateAvg = candyDF.groupby("chocolate")["winpercent"].mean()[1]
fruityAvg = candyDF.groupby("fruity")["winpercent"].mean()[1]
caramelAvg = candyDF.groupby("caramel")["winpercent"].mean()[1]
nutsAvg = candyDF.groupby("peanutyalmondy")["winpercent"].mean()[1]
```

```python
ingredients = ["Chocolate", "Fruity", "Caramel", "Peanut/Almond"]
averages = [chocolateAvg, fruityAvg, caramelAvg, nutsAvg]

plt.bar(ingredients, averages, color = "#9f6db8")
plt.xlabel("Ingredient Type")
plt.ylabel("Average Win Percent")
plt.title("Average Popularity by Ingredient Type")
plt.xticks(rotation=45)
plt.show()
```

Figure 4: Pluribus vs Popularity

```python
pluribusAvg = candyDF.groupby("pluribus")["winpercent"].mean()
barColors = ["#8263a8", "#b2adb3"]

plt.bar(["Single", "Pluribus"], pluribusAvg, color = barColors)
plt.xlabel("Candy Type")
plt.ylabel("Average Win Percent")
plt.title("Average Popularity: Pluribus vs Single")
plt.show()
```

Figure 5: Price vs Popularity

```python
plt.scatter(candyDF["pricepercent"], candyDF["winpercent"], c = "#b67abf", alpha = 0.5)
plt.xlabel("Price Percent")
plt.ylabel("Win Percent")
plt.title("Relationship Between Price and Popularity")
plt.show()
```

Figure 6: Sugar vs Popularity

```python
plt.scatter(candyDF["sugarpercent"], candyDF["winpercent"], c = "#b67abf", alpha = 0.5)
plt.xlabel("Sugar Percent")
plt.ylabel("Win Percent")
plt.title("Relationship Between Sugar Content and Popularity")
plt.show()
```
